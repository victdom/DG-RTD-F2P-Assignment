# F2P Project Deliverables & Metadata Exploration Demo
Demonstration of basic data exploration for DG RTD.H5 policy feedback services.

```python
import pandas as pd

# Create a small simulated dataset of Horizon Europe research projects
data = {
    'Project_ID': ['HE-001', 'HE-002', 'HE-003', 'HE-004'],
    'Project_Title': ['Clean Hydrogen Transition', 'AI in Healthcare', 'Sustainable Agri-Food', 'Solar Storage Tech'],
    'Policy_Cluster': ['Climate & Energy', 'Digital & Health', 'Bioeconomy', 'Climate & Energy'],
    'Status': ['Completed', 'Ongoing', 'Completed', 'Completed'],
    'Budget_EUR': [2500000, 4100000, 1800000, 3200000]
}

df = pd.DataFrame(data)
df
```

```python
# Filter projects related to 'Climate & Energy' for policy review
climate_projects = df[(df['Policy_Cluster'] == 'Climate & Energy') & (df['Status'] == 'Completed')]
print("Projects available for Green Deal Policy Briefings:")
climate_projects
```

```python
# Calculate aggregate statistics for policy briefing notes
total_budget = climate_projects['Budget_EUR'].sum()
project_count = len(climate_projects)
print(f"Total Completed Projects: {project_count}")
print(f"Total Invested Budget: €{total_budget:,.2f}")
```
