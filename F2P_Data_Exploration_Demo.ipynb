# F2P Project Deliverables & Metadata Exploration Demo
## Demonstration of basic data exploration for DG RTD.H5 policy feedback services.

### **Step 1**: Dataset Creation

```python
import pandas as pd

# Create a small simulated dataset of Horizon Europe research projects
data = {
    'Project_ID': ['HE-001', 'HE-002', 'HE-003', 'HE-004'],
    'Project_Title': ['Clean Hydrogen Transition', 'AI in Healthcare', 'Sustainable Agri-Food', 'Solar Storage Tech'],
    'Policy_Cluster': ['Climate & Energy', 'Digital & Health', 'Bioeconomy', 'Climate & Energy'],
    'Status': ['Completed', 'Ongoing', 'Completed', 'Completed'],
    'Budget_EUR': [2500000, 4100000, 1800000, 3200000]
}

df = pd.DataFrame(data)
df
```

### **Step 2**: Filter Policy Data

```python
# Filter projects related to 'Climate & Energy' for policy review
climate_projects = df[(df['Policy_Cluster'] == 'Climate & Energy') & (df['Status'] == 'Completed')]
print("Projects available for Green Deal Policy Briefings:")
climate_projects
```

### **Step 3**: Summary Statistics

```python
# Calculate aggregate statistics for policy briefing notes
total_budget = climate_projects['Budget_EUR'].sum()
project_count = len(climate_projects)
print(f"Total Completed Projects: {project_count}")
print(f"Total Invested Budget: €{total_budget:,.2f}")
```

### **Step 4**: Visual Portfolio Overview (Budget by Policy Cluster)
This chart illustrates how research funding is distributed across policy domains, supporting rapid evidence synthesis for policy briefings.

```python
import matplotlib.pyplot as plt

# Group total budget by Policy Cluster
cluster_summary = df.groupby('Policy_Cluster')['Budget_EUR'].sum() / 1_000_000  # Convert to Millions

# Create a clean horizontal bar chart
plt.figure(figsize=(8, 3.5))
bars = plt.barh(cluster_summary.index, cluster_summary.values, color='#003366', height=0.5)

# Add titles and labels in plain language
plt.title('Total Research Funding by Policy Cluster (€ Millions)', fontsize=14, fontweight='bold', color='#003366')
plt.xlabel('Budget (€ Millions)', fontsize=10)
plt.xlim(0, 7)
plt.grid(axis='x', linestyle='--', alpha=0.5)

# Add exact value labels next to each bar
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.15, bar.get_y() + bar.get_height()/2, f'€{width:.1f}M', va='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Key Takeaways for Policy Teams
* **Portfolio Overview:** Climate & Energy holds the largest funding share (€5.7M across two completed projects: *Clean Hydrogen Transition* at €2.5M and *Solar Storage Tech* at €3.2M), followed by Digital & Health (€4.1M) and Bioeconomy (€1.8M).
* **Policy Readiness:** Filtering by cluster and completion status immediately isolates verified, ready-to-use research deliverables for upcoming policy briefings.
* **Next Step for AI Integration:** These filtered project records can now be fed into a corporate LLM prototype to generate plain-language summaries with direct links back to the source reports.
